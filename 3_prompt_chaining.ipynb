```python
from langgraph.graph import StateGraph , START , END
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
from typing import TypedDict
```

```python
load_dotenv()

model=ChatGoogleGenerativeAI(model='gemini-3.1-flash-lite')
```

```python
class BlogState(TypedDict):
    title: str
    outline:str
    content: str
```

```python
def create_outline(state: BlogState) -> BlogState:

    # fetch title
    title=state['title']

    # call llm and generate outlint
    prompt=f"Generate an outline for a blog on the topic {title}"
    response=model.invoke(prompt).content

    # update state
    state['outline']=response[0]['text'] if isinstance(response, list) else response

    return state
```

```python
def create_blog(state: BlogState) -> BlogState:

    # fetch title and outline
    title=state['title']
    outline=state['outline']

    # call llm and generate outlint
    prompt=f"Write a blog on the title {title} with the following outline \n {outline}"
    response=model.invoke(prompt).content

    # update state
    state['content']=response[0]['text'] if isinstance(response, list) else response

    return state
```

```python
graph=StateGraph(BlogState)

# add nodes
graph.add_node('create_outline',create_outline)
graph.add_node('create_blog',create_blog)

# add edges
graph.add_edge(START,'create_outline')
graph.add_edge('create_outline','create_blog')
graph.add_edge('create_blog',END)


workflow=graph.compile()
```

```python
initial_state={'title':'Rise of AI '}

final_satte=workflow.invoke(initial_state)
```
